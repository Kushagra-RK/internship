## Feature engineering

### Subtask:
Create new features from existing ones to potentially improve the model's performance. This step can be customized based on domain knowledge.


**Reasoning**:
Create new features by performing arithmetic operations on existing columns, handling potential division by zero.



```python
# Create new feature 'Flow_Packets_per_Second', handling division by zero
df['Flow_Packets_per_Second'] = df.apply(
    lambda row: row[' Flow Packets/s'] / row[' Flow Duration'] if row[' Flow Duration'] != 0 else 0, axis=1
)

# Create new feature 'Avg_Packet_Size_Flow', handling division by zero
df['Avg_Packet_Size_Flow'] = df.apply(
    lambda row: row['Flow Bytes/s'] / row[' Flow Packets/s'] if row[' Flow Packets/s'] != 0 else 0, axis=1
)

# Create new feature 'Packet_Length_to_Flow_Duration', handling division by zero
df['Packet_Length_to_Flow_Duration'] = df.apply(
    lambda row: row[' Packet Length Mean'] / row[' Flow Duration'] if row[' Flow Duration'] != 0 else 0, axis=1
)

# Create new feature 'Fwd_Bwd_Packet_Ratio', handling division by zero
df['Fwd_Bwd_Packet_Ratio'] = df.apply(
    lambda row: row[' Total Fwd Packets'] / row[' Total Backward Packets'] if row[' Total Backward Packets'] != 0 else 0, axis=1
)

# Create new feature 'Fwd_Bwd_Byte_Ratio', handling division by zero
df['Fwd_Bwd_Byte_Ratio'] = df.apply(
    lambda row: row['Total Length of Fwd Packets'] / row[' Total Length of Bwd Packets'] if row[' Total Length of Bwd Packets'] != 0 else 0, axis=1
)

display(df[['Flow_Packets_per_Second', 'Avg_Packet_Size_Flow', 'Packet_Length_to_Flow_Duration', 'Fwd_Bwd_Packet_Ratio', 'Fwd_Bwd_Byte_Ratio']].head())
```

## Apply pca

### Subtask:
Reduce the dimensionality of the dataset using PCA while retaining a significant amount of the variance.


**Reasoning**:
Instantiate a PCA object, apply PCA to the scaled numerical data, and store the principal components in a new DataFrame.



```python
from sklearn.decomposition import PCA
import numpy as np

# Check for NaNs and infinite values in df_scaled before PCA
if df_scaled.isnull().sum().sum() > 0 or np.isinf(df_scaled).sum().sum() > 0:
    print("df_scaled contains NaN or infinite values. Please check the scaling step.")
else:
    pca = PCA(n_components=0.95)
    df_pca = pca.fit_transform(df_scaled)

    df_pca = pd.DataFrame(df_pca)
    display(df_pca.head())
```

## Visualize pca results

### Subtask:
Visualize the results of PCA, perhaps using a scatter plot if the reduced dimensionality is 2 or 3, to inspect the separation of different classes.


**Reasoning**:
Check the number of components in df_pca to determine if a scatter plot visualization is feasible.



```python
n_components = df_pca.shape[1]
if n_components == 2:
    plt.figure(figsize=(10, 8))
    scatter = plt.scatter(df_pca.iloc[:, 0], df_pca.iloc[:, 1], c=df[' Label'].astype('category').cat.codes, cmap='viridis', s=10)
    plt.xlabel('Principal Component 1')
    plt.ylabel('Principal Component 2')
    plt.title('PCA of Network Traffic Data (2 Components)')
    legend = plt.legend(*scatter.legend_elements(), title="Labels")
    plt.show()
elif n_components == 3:
    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')
    scatter = ax.scatter(df_pca.iloc[:, 0], df_pca.iloc[:, 1], df_pca.iloc[:, 2], c=df[' Label'].astype('category').cat.codes, cmap='viridis', s=10)
    ax.set_xlabel('Principal Component 1')
    ax.set_ylabel('Principal Component 2')
    ax.set_zlabel('Principal Component 3')
    plt.title('PCA of Network Traffic Data (3 Components)')
    legend = ax.legend(*scatter.legend_elements(), title="Labels")
    plt.show()
else:
    print(f"Scatter plot visualization is not feasible as the number of components ({n_components}) is greater than 3.")
```

Extract the top 10 most important features which causes the attack using Random Forest (preferred) or Decision Tree algorithms. In Random Forest, set n_estimators = 50, as it performs efficiently on large datasets (over 50,000 rows).

## Prepare data

### Subtask:
Separate the features (X) and the target variable (y).


**Reasoning**:
Separate the features (X) and the target variable (y) by dropping the ' Label' column and selecting the ' Label' column respectively.



```python
X = df.drop(' Label', axis=1)
y = df[' Label']
display(X.head())
display(y.head())
```

## Train random forest model

### Subtask:
Train a Random Forest classifier on the data.


**Reasoning**:
Train a Random Forest classifier on the data.



```python
from sklearn.ensemble import RandomForestClassifier

# Instantiate a RandomForestClassifier object with n_estimators=50 and random_state=42
rf_model = RandomForestClassifier(n_estimators=50, random_state=42)

# Fit the Random Forest model to the features X and the target variable y
rf_model.fit(X, y)
```

## Extract feature importances

### Subtask:
Get the feature importances from the trained model.


**Reasoning**:
Get the feature importances from the trained model and store them in a pandas Series with the correct index.



```python
feature_importances = rf_model.feature_importances_
feature_importance_series = pd.Series(feature_importances, index=X.columns)
display(feature_importance_series.head())
```

## Select top features

### Subtask:
Select the top 10 most important features based on the importances.


**Reasoning**:
Sort the feature importances and select the top 10 feature names.



```python
top_10_features = feature_importance_series.sort_values(ascending=False).head(10).index.tolist()
print("Top 10 most important features:")
print(top_10_features)
```

## Visualize top features

### Subtask:
Visualize the top 10 most important features to gain insights into their relative importance.


**Reasoning**:
Create a bar plot to visualize the top 10 features' importances and order the bars by importance in descending order.



```python
import matplotlib.pyplot as plt
import seaborn as sns

top_10_importance = feature_importance_series.sort_values(ascending=False).head(10)

plt.figure(figsize=(12, 6))
sns.barplot(x=top_10_importance.index, y=top_10_importance.values)
plt.title('Top 10 Most Important Features')
plt.xlabel('Features')
plt.ylabel('Importance')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
